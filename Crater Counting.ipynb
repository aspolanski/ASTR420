```python
import numpy as np
import matplotlib.pyplot as plt
import smplotlib
from itertools import product
```

## Part a) Model 1: Uniform arrangement of craters. 0.5 pts

```python
fig, ax = plt.subplots(figsize=(8,6))

ax.set_ylim(0,6)
ax.set_xlim(0,8)
ax.grid()

x = np.arange(0,8,1)
y = np.arange(0,6,1)

for xi, yi in product(x, y):
    
    crater = plt.Circle((xi+np.random.uniform(0,1,1),
                         yi+np.random.uniform(0,1,1)), 
                        0.3, edgecolor='k',facecolor='grey',alpha=0.5)

    ax.add_patch(crater)
    
```

## Part a) Model 2: Random arrangement of craters. 0.5 pts

```python
fig, ax = plt.subplots(figsize=(8,6))

ax.set_ylim(0,6)
ax.set_xlim(0,8)
ax.grid()

coords = np.random.uniform(0,1,(2,48)) * np.array([8,6])[:,np.newaxis]

for i in range(48):
    
    crater = plt.Circle((coords[0,i], coords[1,i]), 0.3, edgecolor='k',facecolor='grey',alpha=0.5)

    ax.add_patch(crater)
```

## Part b) How does the evenly spaced model compare to the random models? Were you surprised by how a “random” surface looks? If so, how? Comment on the tendency to perceive structure in the random maps. 1 pt.

### Both Model 1 and 2 look similar but the random model has a greater tendency to have groups of craters. These groups can be interpreted as regions of older surface even though the cratering rate was just random. Humans have an inate ability to discern patterns from essentially random noise - an important bias that needs to be kept in mind when interpreting crater maps.

### Separating the two surfaces really only becomes possible past 100-200% enhancement. In the other cases, one can justify various boundaries other than the halfway division.
