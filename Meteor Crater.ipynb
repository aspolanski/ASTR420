```python
import numpy as np
import matplotlib.pyplot as plt
import smplotlib


def crater_R(rhoi,rhop,Ri,vi):
    
    return 1.103 * (rhoi/rhop)**0.33 * Ri**0.78 * 9.8**-0.22 * vi**0.44


rhoi = 7800
rhop = 2600
```

```python
vis = np.linspace(9000,40000,1000)
ris = np.linspace(1,75,1000)

V, R = np.meshgrid(vis,ris)


fig, ax = plt.subplots()
Rs = crater_R(rhoi,rhop,R,V)

cs = ax.contourf(V/1000, R, Rs, levels=20, cmap="coolwarm")

cs_lines = ax.contour(V/1000, R, Rs, levels=cs.levels, colors="black", linewidths=0.5)
ax.clabel(cs_lines, inline=True, fontsize=8, fmt="%.2f")

ax.set_ylabel("Impactor Radius [m]")
ax.set_xlabel(r"Impactor Velocity [kms$^{-1}$]")
```
